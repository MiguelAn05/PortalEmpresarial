## 2. Valores y tipos

Todo lo que maneja un programa es un **valor**, y cada valor tiene un **tipo**: la
clase de cosa que es. El tipo decide qué se puede hacer con él (un número se puede
dividir; un texto no).

| Tipo | Qué guarda | Ejemplo | En el portal |
|---|---|---|---|
| `int` | Números enteros | `15` | Los días de plazo |
| `float` | Números con decimales | `2.3` | El promedio por día hábil del informe |
| `str` | Texto (*string*) | `"reclamo"` | El tipo, el estado, la descripción |
| `bool` | Verdadero o falso | `True`, `False` | ¿Está terminada? ¿Es institucional? |
| `None` | **Nada**: no hay valor | `None` | Una PQRS sin fecha de cierre |

Cuidado con dos detalles:

- Los decimales se escriben con **punto**, no con coma: `2.3`, no `2,3`.
- `True`, `False` y `None` van con la primera letra en **mayúscula**. `true` sería un
  error.

`type(valor)` te dice de qué tipo es algo. Corre la celda y mira las cinco respuestas.

````python
print(type(15))
print(type(2.3))
print(type("reclamo"))
print(type(True))
print(type(None))
````

## 3. Variables: ponerle nombre a un valor

Una **variable** es un nombre que apunta a un valor, para poder usarlo después. Se
crea con `=`:

```python
dias = 8
```

Esto se lee **«`dias` vale 8»**, no «dias es igual a 8». El `=` **guarda**; no compara
(para comparar se usa `==`, sección 4).

Reglas de los nombres:

- Letras, números y guion bajo `_`. Sin espacios, sin tildes ni `ñ` (Python las
  acepta, pero el portal las evita por costumbre), y sin empezar con número.
- Distingue mayúsculas: `dias` y `Dias` son dos variables distintas.
- Por convención, las palabras se separan con `_`: `fecha_limite`, `area_responsable`.

**Dos costumbres del portal:**

- **Los nombres van en español** (regla de CLAUDE.md: *«El código se lee como el
  dominio»*). Por eso vas a leer `solicitud`, `estado`, `area_causante`, no `request`,
  `status`, `cause_area`.
- **Lo que no debe cambiar se escribe en MAYÚSCULAS**: `SLA_DIAS_POR_TIPO`,
  `MAX_PRODUCTOS`. Python no lo impide, pero quien lee entiende «esto es una regla fija,
  no lo toques en medio del programa». Se les llama **constantes**.

En un notebook, la **última línea** de una celda se muestra sola, sin `print`.

````python
tipo = "reclamo"
dias = 8
prioridad = "alta"
terminada = False
fecha_cierre = None

dias
````

Una variable se puede **cambiar**: se le asigna otro valor y el anterior se pierde.
Aquí aparece también `+=`, que es un atajo: `dias += 2` significa «`dias` vale lo que

````python
dias = 8
dias = dias + 2      # se lee: «dias vale lo que valía, más 2»
print(dias)
dias += 5            # lo mismo, más corto
print(dias)
````

## 4. Operaciones

### Con números

| Símbolo | Qué hace | Ejemplo | Resultado |
|---|---|---|---|
| `+` `-` `*` | Suma, resta, multiplicación | `8 * 2` | `16` |
| `/` | División: **siempre** da decimal | `15 / 2` | `7.5` |
| `//` | División entera: sin decimales | `15 // 2` | `7` |
| `%` | El **sobrante** de la división | `15 % 2` | `1` |
| `round(x, n)` | Redondea a `n` decimales | `round(2.3456, 2)` | `2.35` |

El informe de PQRS usa exactamente esto para el promedio por día hábil:
`round(n / dias, 2)` — divide y deja dos decimales.

````python
pqrs_del_mes = 30
dias_habiles = 13
promedio = round(pqrs_del_mes / dias_habiles, 2)
promedio
````

### Con texto: los f-strings

Para armar un texto con valores adentro, el portal usa **f-strings**: un texto que
empieza con `f` antes de las comillas, y donde lo que va entre `{ }` se reemplaza por
su valor. Aparecen en todos los mensajes de error y en todos los correos.

Este mensaje es, palabra por palabra, el que el portal le muestra a quien intenta
cerrar una PQRS sin tener permiso (`pqrs/gestion.py`):

````python
quien = "Servicio al Cliente"
print(f"Una PQRS la cierra {quien}. Márcala como 'resuelto' y la revisan y la cierran.")
````

Dentro de las llaves puede ir una cuenta, y después de `:` un **formato**. `{n:04d}`
escribe el número con 4 dígitos rellenando con ceros: así nace el «0042» del código
`PVG0042`.

````python
prefijo = "PVG"
numero = 42
f"{prefijo}{numero:04d}"
````

### Comparar: el resultado siempre es `True` o `False`

| Símbolo | Pregunta |
|---|---|
| `==` | ¿Son iguales? |
| `!=` | ¿Son distintos? |
| `<` `>` `<=` `>=` | ¿Menor, mayor…? |

**`=` guarda; `==` pregunta.** Confundirlos es el error más común del principio.

````python
estado = "cerrado"
print(estado == "cerrado")       # ¿es cerrado?
print(estado != "anulado")       # ¿es distinto de anulado?
print(8 < 15)                    # ¿el plazo de un reclamo es menor que el de una petición?
print("Reclamo" == "reclamo")    # las mayúsculas cuentan
````

Esa última línea es la razón de varias funciones del portal que pasan todo a minúsculas
antes de comparar (`sinTildes`, `normalizar`): para Python, «Reclamo» y «reclamo» son
textos distintos, y un buscador que no lo tenga en cuenta no encuentra nada.

## 5. Listas y tuplas: varios valores en orden

Una **lista** guarda varios valores, en orden, entre corchetes `[ ]` y separados por
comas.

- Cada elemento tiene una **posición** (*índice*) que **empieza en 0**, no en 1:
  `estados[0]` es el primero. Es lo más raro del principio y te va a confundir unas
  cuantas veces; a todo el mundo le pasa.
- `estados[-1]` es el último (los negativos cuentan desde el final).
- `len(lista)` dice cuántos elementos tiene (*length*, largo).
- `valor in lista` pregunta si está: `True` o `False`.

````python
estados = ["recibido", "asignado", "en_proceso", "resuelto", "cerrado"]

print(estados[0])          # el primero: posición 0
print(estados[2])          # el tercero
print(estados[-1])         # el último
print(len(estados))        # cuántos hay
print("cerrado" in estados)
print("anulado" in estados)
````

Una lista se puede **cambiar**: `lista.append(valor)` le agrega uno al final. Esto
pasó de verdad en el portal hace unos días, cuando se creó el estado «anulado»:

````python
estados.append("anulado")
estados
````

### Tuplas: listas que no cambian

Una **tupla** es como una lista pero con paréntesis `( )`, y **no se puede cambiar**
después de creada. El portal las usa para las reglas fijas. Esta es real, de
`backend/app/modules/pqrs/pendientes.py`, línea 32: los estados en que el plazo de una
PQRS sigue corriendo.

```python
ESTADOS_ABIERTOS = ("recibido", "asignado", "en_proceso")
```

¿Por qué tupla y no lista? Porque si fuera lista, cualquier parte del programa podría
hacer `ESTADOS_ABIERTOS.append(...)` sin querer y cambiar la regla para todos. Con una
tupla, Python no lo deja:

````python
ESTADOS_ABIERTOS = ("recibido", "asignado", "en_proceso")

print("resuelto" in ESTADOS_ABIERTOS)     # ¿una resuelta tiene el plazo corriendo?
ESTADOS_ABIERTOS.append("resuelto")        # esto va a fallar, a propósito
````

`AttributeError: 'tuple' object has no attribute 'append'`: *«una tupla no sabe
agregar»*. Ese error es justo la protección que se buscaba.

Y la primera línea respondió algo de negocio: **una PQRS resuelta ya no tiene el plazo
corriendo** (ya se le respondió al cliente; solo falta que confirme).

## 6. Diccionarios: valores con nombre

Una lista guarda valores por **posición** (0, 1, 2…). Un **diccionario** los guarda
por **nombre**: cada elemento es un par **clave → valor**, entre llaves `{ }`, con
dos puntos entre la clave y el valor.

Es exactamente la forma de las trece líneas del principio: a cada tipo de PQRS (la
clave) le corresponde un número de días (el valor).

- `diccionario[clave]` trae el valor de esa clave.
- `diccionario[clave] = valor` agrega o cambia una.

````python
SLA_DIAS_POR_TIPO = {
    "peticion":   15,
    "queja":       5,
    "reclamo":     8,
    "sugerencia": 10,
}

print(SLA_DIAS_POR_TIPO["reclamo"])
print(SLA_DIAS_POR_TIPO["peticion"])
print(len(SLA_DIAS_POR_TIPO), "tipos con plazo")
````

Los espacios de más después de los dos puntos (`"queja":       5`) no significan
nada para Python: alguien los puso para que los números quedaran **alineados** y se
leyeran como una tabla. Es una costumbre del portal: el código también se escribe para
que lo lea una persona.

### Una clave que no existe: `KeyError`

¿Y la felicitación? No está en el diccionario. Pedirla con corchetes **revienta**:

````python
SLA_DIAS_POR_TIPO["felicitacion"]
````

`KeyError: 'felicitacion'`: *«no hay ninguna clave con ese nombre»*.

### `.get(clave, por_defecto)`: preguntar sin reventar

`.get()` hace lo mismo que los corchetes, pero si la clave no existe devuelve el
segundo valor, **el valor por defecto**, en vez de fallar. Es como decir «dame los
días del tipo; si no lo tienes anotado, usa 10».

````python
print(SLA_DIAS_POR_TIPO.get("reclamo", 10))        # existe: 8
print(SLA_DIAS_POR_TIPO.get("felicitacion", 10))   # no existe: el valor por defecto
print(SLA_DIAS_POR_TIPO.get("felicitacion"))       # sin valor por defecto: None
````

Ahí está, adelantada, la respuesta a la pregunta de la felicitación. La confirmamos en
la sección 9 con el código de verdad.

### Recorrer un diccionario: `.items()`

`.items()` entrega los pares clave-valor, uno por uno. Para usarlo hace falta un `for`,
que llega en la sección 8; por ahora mira solo el resultado.

````python
for tipo, dias in SLA_DIAS_POR_TIPO.items():
    print(tipo, "→", dias, "días hábiles")
````

## 7. Decisiones: `if`, `elif`, `else`

Para que el programa haga una cosa **u otra** según una condición:

```python
if condicion:
    lo que pasa si es verdadera
elif otra_condicion:
    lo que pasa si la primera no, pero esta sí
else:
    lo que pasa si ninguna
```

Tres reglas que causan casi todos los errores al principio:

1. La línea del `if` **termina en dos puntos** `:`.
2. Lo que depende del `if` va **indentado**: corrido 4 espacios a la derecha. En Python
   la indentación **no es decoración**: es lo que dice qué está dentro del `if` y qué
   no.
3. `elif` y `else` son opcionales. Se revisan **en orden**, y solo entra al primero que
   se cumpla.

````python
estado = "resuelto"

if estado == "cerrado":
    print("Ya terminó: se le mandó la encuesta al cliente.")
elif estado == "resuelto":
    print("Se le respondió al cliente: falta que confirme.")
else:
    print("Sigue abierta: el plazo está corriendo.")
````

Cambia `"resuelto"` por `"en_proceso"` o por `"cerrado"` en la primera línea y vuelve
a correr la celda. Así se aprende: cambiando una cosa y mirando qué pasa.

### Combinar condiciones: `and`, `or`, `not`

| Palabra | Es verdadero cuando… |
|---|---|
| `a and b` | las **dos** se cumplen |
| `a or b` | **al menos una** se cumple |
| `not a` | `a` **no** se cumple |

Esta es una regla real de la sección de notas crédito: la bodega solo se pregunta si la
venta es institucional **y** el motivo tiene que ver con producto.

````python
institucional = True
motivo_con_producto = False

if institucional and motivo_con_producto:
    print("Hay que preguntar de qué bodega salió el producto.")
else:
    print("No se pregunta bodega.")
````

### Lo «vacío» cuenta como falso

En un `if`, Python trata como **falso** todo lo que está vacío: `""` (texto vacío),
`0`, `None`, `[]` (lista vacía), `{}`. Todo lo demás cuenta como verdadero.

El portal lo usa todo el tiempo. Esta línea es real, de `pqrs/gestion.py`:

```python
if estado == "resuelto" and not (solucion or "").strip():
```

Parece jeroglífico, pero ya tienes casi todas las piezas. Se lee de adentro hacia
afuera:

| Pedazo | Qué hace |
|---|---|
| `solucion or ""` | Si `solucion` es `None`, usa `""` (texto vacío) en su lugar. Así lo que sigue no revienta |
| `.strip()` | Le quita los espacios de las puntas: `"   "` queda `""` |
| `not (...)` | «¿Quedó vacío?» |
| `estado == "resuelto" and ...` | «¿La están pasando a resuelto **y** no escribieron la solución?» |

En palabras: **no se puede marcar una PQRS como resuelta sin escribir qué se le
solucionó al cliente**, y unos espacios en blanco no cuentan como haber escrito algo.

````python
for solucion in ["Se cambió el producto", "   ", "", None]:
    falta = not (solucion or "").strip()
    print(repr(solucion), "→ ¿falta la solución?", falta)
````

`repr(valor)` muestra el valor **como se escribiría en el código**, con comillas: así
se ven los espacios y la diferencia entre `''` y `None`.

Fíjate en la tercera y cuarta línea: sin el `or ""`, el `None` habría reventado con
`AttributeError` (*«None no sabe hacer strip»*). Ese truco protege contra un campo que
el formulario no mandó.

## 8. Repeticiones: `for`

`for elemento in coleccion:` repite el bloque indentado **una vez por cada elemento**,
y en cada vuelta `elemento` vale el siguiente. Sirve con listas, tuplas, diccionarios,
textos…

Igual que el `if`: dos puntos al final e indentación adentro.

````python
tipos = ["peticion", "queja", "reclamo"]

for tipo in tipos:
    print("Revisando:", tipo)

print("Listo")     # sin indentación: está FUERA del for, corre una sola vez al final
````

### Contar: el acumulador

El patrón más común: una variable que empieza en 0 y en cada vuelta suma. Así cuenta
el informe de PQRS cuántas hay de cada cosa.

````python
estados_de_hoy = ["recibido", "cerrado", "en_proceso", "cerrado", "resuelto", "cerrado"]

cerradas = 0
for estado in estados_de_hoy:
    if estado == "cerrado":
        cerradas += 1

print(cerradas, "cerradas de", len(estados_de_hoy))
````

### `range` y `enumerate`

- `range(n)` da los números de `0` a `n-1`. Útil para repetir algo `n` veces.
- `enumerate(lista, start=1)` da **la posición y el elemento** juntos.

````python
for numero in range(1, 4):           # del 1 al 3: el último NO se incluye
    print(f"PVG{numero:04d}")

for posicion, estado in enumerate(["recibido", "asignado"], start=1):
    print(posicion, estado)
````

### Una lista armada en una sola línea

El portal tiene muchas líneas como esta, del informe de PQRS
(`pqrs/informe.py`, línea 148):

```python
solicitudes = [s for s in todas if s.estado != "anulado"]
```

Es un `for` con un `if`, escrito en una línea. Se llama **comprensión de lista** y se
lee así: **«de todas, quédate con cada `s` cuyo estado no sea anulado»**. Las dos
formas de la celda hacen exactamente lo mismo:

````python
todas = ["recibido", "anulado", "cerrado", "anulado", "en_proceso"]

# La forma larga
sin_anuladas = []
for estado in todas:
    if estado != "anulado":
        sin_anuladas.append(estado)
print(sin_anuladas)

# La forma corta, como en el portal
print([estado for estado in todas if estado != "anulado"])
````

## 9. El código real, línea por línea

Ya tienes todo lo necesario. Abrimos el archivo de verdad. Para no depender de números
de línea que se pueden correr, usamos una pequeña herramienta, `buscar()` (la misma del
nivel 0). Todavía no sabes escribir funciones (`def` llega en el notebook 2); por ahora
úsala como una caja negra: le das un archivo y un texto, y te muestra las líneas.

````python
from pathlib import Path

PORTAL = Path.cwd().parent / "protokimica-portal"

def buscar(archivo, texto, despues=0):
    "Muestra las líneas que contienen `texto` y, si se pide, las `despues` que siguen."
    lineas = (PORTAL / archivo).read_text(encoding="utf-8").splitlines()
    for i, linea in enumerate(lineas):
        if texto in linea:
            for j in range(i, min(i + despues + 1, len(lineas))):
                print(f"{j + 1:>4}  {lineas[j]}")
            print()

buscar("backend/app/modules/pqrs/service.py", "SLA_DIAS_POR_TIPO = {", despues=12)
````

Línea por línea:

| Línea | Qué dice | Qué significa |
|---|---|---|
| `SLA_DIAS_POR_TIPO = {` | Crea una variable con un diccionario | Nombre en MAYÚSCULAS: es una **constante**, una regla fija. «SLA» es *acuerdo de nivel de servicio*: el plazo de respuesta |
| `"peticion":   15,` | La clave `"peticion"` vale `15` | Una petición tiene 15 días hábiles: lo dice la **Ley 1755 de 2015** |
| `"queja":       5,` | `"queja"` vale `5` | Una queja, 5 días |
| `"reclamo":     8,` | `"reclamo"` vale `8` | Un reclamo, 8 |
| `"sugerencia": 10,` | `"sugerencia"` vale `10` | Una sugerencia, 10 |
| `}` | Cierra el diccionario | — |
| *(línea vacía)* | — | Separa una regla de la otra, para leer |
| `PRIORIDAD_POR_TIPO = {` | Otro diccionario: el tipo da la prioridad | Clave → texto, en vez de número |
| `"queja": "alta"`, `"reclamo": "alta"` | — | Quejas y reclamos, prioridad alta |
| `"sugerencia": "baja"` | — | Una sugerencia no apura a nadie |

Las claves están escritas **sin tilde** (`"peticion"`, no `"petición"`). No es un
descuido: son claves internas que también se guardan en la base de datos, y sin tildes
se evitan problemas al compararlas. Lo que ve la gente («Petición») se escribe aparte,
en la pantalla.

### ¿Quién usa estos diccionarios?

Dos funciones del mismo archivo. Todavía no sabes escribir funciones, pero ya puedes

4. ¿Cuál es el promedio por día hábil?

````python
pqrs_del_mes = [
    {"codigo": "PVG0040", "tipo": "reclamo",      "estado": "cerrado"},
    {"codigo": "PVG0041", "tipo": "queja",        "estado": "en_proceso"},
    {"codigo": "PVG0042", "tipo": "reclamo",      "estado": "recibido"},
    {"codigo": "VI0036",  "tipo": "peticion",     "estado": "resuelto"},
    {"codigo": "VI0037",  "tipo": "peticion",     "estado": "anulado"},
    {"codigo": "PVB0010", "tipo": "felicitacion", "estado": "cerrado"},
    {"codigo": "PVB0011", "tipo": "reclamo",      "estado": "asignado"},
]
DIAS_HABILES = 13

print(pqrs_del_mes[2])              # la tercera PQRS: un diccionario
print(pqrs_del_mes[2]["codigo"])    # su código: primero la posición, después la clave
````

````python
# 1. Sin las anuladas
reales = [p for p in pqrs_del_mes if p["estado"] != "anulado"]
print("Radicadas:", len(reales), "(sin contar", len(pqrs_del_mes) - len(reales), "anulada)")

# 2. Cuántas de cada tipo: un diccionario que va contando
por_tipo = {}
for p in reales:
    por_tipo[p["tipo"]] = por_tipo.get(p["tipo"], 0) + 1
print("Por tipo:", por_tipo)

# 3. Las que tienen el plazo corriendo
con_plazo = [p["codigo"] for p in reales if p["estado"] in ESTADOS_ABIERTOS]
print("Con el plazo corriendo:", con_plazo)

# 4. El promedio por día hábil
print("Por día hábil:", round(len(reales) / DIAS_HABILES, 2))
````

por_tipo[p["tipo"]] = por_tipo.get(p["tipo"], 0) + 1

Léela de derecha a izquierda: «cuántas llevo de este tipo (**0 si es el primero**), más
